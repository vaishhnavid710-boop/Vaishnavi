```python
# nested loop
List_1=['Red','Yellow','Lightgreen']
List_2= ['White','Black','Yellow']
List_3= ['White','Purple','Yellow']

for i in List_1:
    for j in List_2:
        for k in List_3:
            if i==j==k:
                print('Matched',i,j,k)
```
