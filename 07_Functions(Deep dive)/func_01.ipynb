```python
List_01 = ['1','2','3','4','5']
def myDecorator(func):
    def wrapper(List_01):
        int_list = [int(i) for i in List_01]
        print(int_list)
        result = func(int_list)
        print(result)
        output = dict(zip(int_list,result))
        print(output)
    return wrapper

@myDecorator
def originalFunc(List_01):
    newList = []
    for i in List_01:
        newList.append(i**2)
        return newList
    originalFunc(List_01)   
```
